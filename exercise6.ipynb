```python
#실습1
def sum_even(*args):
    if len(args) == 0:
        return -1

    even_sum = 0
    for num in args:
        if num % 2 == 0:
            even_sum += num
        return even_sum

print(sum_even(1, 2, 3, 4, 5))
print(sum_even(1, 3, 5))
print(sum_even())
```

```python
#실습2
def sum_all(*args):
    if len(args) == 0:
        return None
    has_number = False
    total = 0
    for data in args:
        # bool 타입 거르기 -> int 취급
        if type(data) in (int, float):
            total += data
            has_number = True

    return total if has_number else None

print(sum_all(1, 2, "a", 3, True))
print(sum_all("a", "b"))
print(sum_all())
```

```python
#실습3
def make_list(*nums, type=10):
    result = []
    for num in nums:
        result.append(num * type)
    return result

print(make_list(1,2,3,4,5))                       # [ 10, 20, 30, 40, 50 ]
print(make_list(1,2,3,4,5, type=3))               # [ 3, 6, 9, 12, 15 ]
print(make_list(10, 20, 30, 40, 50, 60, type=2))  # [ 20, 40, 60, 80, 100, 120 ]
print(make_list(1,2,3, type=100))                 # [ 100, 200, 300 ]
print(make_list(1,1,1,1,1,1,1, type=0))           # [ 0, 0, 0, 0, 0, 0, 0 ]
```

```python
#실습4
def update_list(nums, inplace=False):
    if inplace:
        for i in range(len(nums)):
            nums[i] = nums[i] ** 2
    else:
        new_list = []
        for num in nums:
            new_list.append(num ** 2)
        return new_list


mynums = [ 3, 7, 2, 6, 10 ]
print(update_list(mynums, False))  # [ 9, 49, 4, 36, 100 ]
print(mynums)                     # [ 3, 7, 2, 6, 10 ] (원본 보존됨)
print(update_list(mynums, True))   # None
print(mynums)                     # [ 9, 49, 4, 36, 100 ] (원본이 제곱으로 바뀜
```

```python
#실습5
def mydict(**kwargs):
    result = {}
    for key, value in kwargs.items():
        result['my' + key] = value
    return result

print(mydict(a=1, b=2))             # {'mya': 1, 'myb': 2}
print(mydict(name="재혁", age=28)) # {'myname': '재혁', 'myage': 28}
print(mydict())                     # {}
```

```python
#실습6
def make_dict(*nums):
        if len(nums) == 0:
            return None

        result = {}
        odd_count = 0
        even_count = 0
        for num in nums:
            if num % 2 == 0:
                even_count += 1
                result[f"even{even_count}"] = num
            else:
                odd_count += 1
                result[f"odd{odd_count}"] = num
        return result
    
print(make_dict(1,2,3,4,5))    # {'odd1': 1, 'even1': 2, 'odd2': 3, 'even2': 4, 'odd3': 5}
print(make_dict(10,23,45,5))   # {'even1': 10, 'odd1': 23, 'odd2': 45, 'odd3': 5}
print(make_dict())             # None
```

```python
#실습7
def myprint(*args, **kwargs):
    if len(args) == 0:
        print("Hello Python")
        return

    deco = kwargs.get('deco', '**')
    sep = kwargs.get('sep', ',')

    str_args = [str(arg) for arg in args]
    content = sep.join(str_args)
    print(f"{deco}{content}{deco}")

# 호출 테스트 예시
myprint(10, 20, 30, deco="@", sep="-")                # @10-20-30@
myprint("python", "javascript", "R", deco="")
myprint("가", "나", "다")                              # **가,나,다**
myprint(100)                                          # **100**
myprint(True, 111, False, "abc", deco="&", sep="")    # &True111Falseabc&
myprint()                                             # Hello Python
```
