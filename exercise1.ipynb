```python
# 실습7
import random

grade = random.randint(1, 6)

if grade >= 1 and grade <= 3:
    print(grade, "학년은 저학년입니다.")
if grade >= 4 and grade <= 6:
    print(grade, "학년은 고학년입니다.")
```

```python
#실습8
grade = random.randint(1, 6)

if grade == 1 or grade == 2 or grade == 3:
    print(grade, "학년은 저학년입니다.")
if grade == 4 or grade == 5 or grade == 6:
    print(grade, "학년은 고학년입니다.")
```

```python
#실습9
score = random.randint(1, 100)

match score // 10:
        case 10 | 9:
            print(score, "점은 A 등급입니다.")
        case 8:
            print(score, "점은 B 등급입니다.")
        case 7:
            print(score, "점은 C 등급입니다.")
        case 6:
            print(score, "점은 D 등급입니다.")
        case _:
            print(score, "점은 F 등급입 니다.")
```
