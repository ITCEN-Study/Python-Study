```python
#실습1
class Member:
    def __init__(self, name, account, passwd, birth_year):
        self.name = name
        self.account = account
        self.passwd = passwd
        self.birth_year = birth_year

# 회원 객체 3개 생성
m1 = Member("원이", "won", "1234", 2004)
m2 = Member("제나", "jena", "5678", 2008)
m3 = Member("미나미", "minami", "abcd", 2006)

print(f"회원 1 : {m1.name}({m1.account},{m1.passwd},{m1.birth_year})")
print(f"회원 2 : {m2.name}({m2.account},{m2.passwd},{m2.birth_year})")
print(f"회원 3 : {m3.name}({m3.account},{m3.passwd},{m3.birth_year})")
```

```python
#실습2
class Member:
    def __init__(self, name, account, passwd, birth_year):
        self.name = name
        self.account = account
        self.passwd = passwd
        self.birth_year = birth_year

    def __str__(self):
        return f"{self.name}({self.account},{self.passwd},{self.birth_year})"

members = [
    Member("원이", "won", "1234", 2004),
    Member("제나", "jena", "5678", 2008),
    Member("미나미", "minami", "abcd", 2006)
]

for idx, member in enumerate(members, 1):
    print(f"회원 {idx} : {member}")
```

```python
#실습3
while True:
    try:
        user_input = input("숫자를 입력하세요 : ")
        limit = int(user_input)

        total_sum = sum(range(1, limit + 1))
        print(f"1부터 {limit}까지의 합은 {total_sum}입니다.")
        break
    except ValueError:
        print("숫자만 입력해 주세요~~~")
```

```python
#실습4
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def getBookInfo(self):
        return f"{self.title}, {self.author}, {self.price}"

    def __str__(self):
        return "Book 클래스로 생성된 객체"

# 책 5개 생성, 리스트 보관
books = [
    Book("포켓몬 생태도감", "주식회사 포켓몬", 15000),
    Book("포켓몬스터 팔데아도감", "학산문화사 편집부", 17000),
    Book("898마리 총 출동! 포켓몬 대도감(상)", "학산문화사 편집부", 17000),
    Book("포켓몬 대도감 1020+", "대원씨아이 편집부", 15000),
    Book("포켓몬스터 과학퀴즈 도감", "서울문화사 편집부", 12000)
]

for idx, book in enumerate(books, 1):
    info = book.getBookInfo().split(",")
    print(f"[ 객체{idx} : {book} ]")
    print(f"책이름 {info[0]}")  # 책이름
    print(f"저자 {info[1]}")  # 저자
    print(f"가격 {info[2]}")  # 가격
    print("*****************************************************")
```

```python
#실습5
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_point(self):
        # 5%
        return int(self.price * 0.05)

    def __str__(self):
        return f"{self.name} 상품의 가격은 {self.price:,} 원입니다."

p1 = Product("둘리인형", 30000)
print(p1)
print(f"포인트 정보 : {p1.get_point()}")

p2 = Product("노트북", 1000000)
print(p2)
print(f"포인트 정보 : {p2.get_point()}")
```

```python
#실습6
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show_info(self):
        print(f"이름은 {self.name}, 급여는 {self.salary:,}원", end="")

class Developer(Employee):
    def __init__(self, name, salary, programming_language):
        super().__init__(name, salary)
        self.programming_language = programming_language

    def show_info(self):
        super().show_info()
        print(f"이며 {self.programming_language} 프로그래밍이 가능합니다.")

developers = [
        Developer("둘리", 150000, "Java"),
        Developer("또치", 120000, "Python"),
        Developer("도우너", 110000, "JavaScript")
]

for dev in developers:
    dev.show_info()
```
