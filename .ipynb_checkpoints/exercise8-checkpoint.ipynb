```python
#실습1
def myemail_info(email):
    if '@' in email:
        return tuple(email.split('@'))
    return None

print(myemail_info("vmfhdirn2@kakao.com"))
print(myemail_info("vmfhdirn2"))
```

```python
#실습2

# (1) s1 = "pythonjavascript" 문자열 길이 출력
s1 = "pythonjavascript"
print(len(s1))

# (2) s2 = "010-7777-9999" 에서 하이픈 제거 후 출력
s2 = "010-7777-9999"
print(s2.replace("-", ""))

# (3) s3 = "PYTHON" 문자열 역순(뒤집기) 출력
s3 = "PYTHON"
print(s3[::-1])

# (4) s4 = "Python" 소문자 변환 출력
s4 = "Python"
print(s4.lower())

# (5) s5 = "https://www.python.org/" 에서 호스트 주소 슬라이싱 추출
s5 = "https://www.python.org/"
print(s5[8:-1])

# (6) 주민등록번호 뒷자리 첫 번째 글자로 성별 판별
s6 = '891022-2473837'
gender_num = s6.split("-")[1][0]
gender = "남자" if gender_num in ["1", "3"] else "여자"
print(gender)

# (7) s7 = '100' 정수 및 실수 형변환 출력
s7 = '100'
print(int(s7))
print(float(s7))

# (8) s8 = ' The Zen of Python' 에서 'n' 문자 개수 출력
s8 = 'The Zen of Python'
print(s8.count('n'))

# (9) s8 에서 'Z' 문자의 인덱스 위치 출력
print(s8.find('Z'))

# (10) s8 대문자 변환 후 공백 기준 분리 리스트 출력
print(s8.upper().split())
```

```python
#실습3
names = ["둘리", "또치", "도우너", "고길동"]
scores = [85, 92, 78, 96]

for name, score in zip(names, scores):
    print(f"{name}: {score}점")
```

```python
#실습4
words = ["apple", "banana", "cherry"]
meanings = ["사과", "바나나", "체리"]

result_dict = dict(zip(words, meanings))
print(result_dict)
```

```python
#실습5
fruits = ["배", "사과", "포도"]
counts = [5, 3, 2]
prices = [3000, 1500, 5000]

for fruit, count, price in zip(fruits, counts, prices):
    print(f"{fruit}의 총가격은 {count * price}원입니다.")
```

```python
#실습6
animals = ["lion", "tiger", "bear", "rabbit", "fox"]
result = []

for idx, animal in enumerate(animals):
    if idx % 2 == 0:
        result.append(animal.upper())
    else:
        result.append(animal)
print(result)
```

```python
#실습7
animals = ["lion", "tiger", "bear", "rabbit", "fox"]

listv2 = [animal.upper() if idx % 2 == 0 else animal for idx, animal in enumerate(animals)]
print(listv2)
```
