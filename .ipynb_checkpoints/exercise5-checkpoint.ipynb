```python
#실습1
total = 0
for i in range(1, 51):
    if i % 3 == 0 and i % 5 !=0:
        total += i
print(f"결과 = {total}")
```

```python
#실습2
total = 0
for i in range(1, 51):
    if i % 3 !=0:
        continue
    if i % 5 == 0:
        continue
    total += i
print(f"결과 = {total} ")
```

```python
#실습3
import random

lotto = []
while len(lotto) < 6:
    num = random.randint(1, 45)
    if num not in lotto:
        lotto.append(num)
print(f"행운의 로또번호 : {lotto}")
```

```python
# 실습4
matrix = [
    [10, 20, 30, 40, 50],
    [5, 10, 15],
    [11, 22, 33, 44],
    [9, 8, 7, 6, 5, 4, 3, 2, 13]
]

for i in range(len(matrix)):
    row_sum = sum(matrix[i])
    print(f"{i+1}행의 합은 {row_sum} 입니다.")
```

```python
# 실습5
matrix = [
    ['B', 'C', 'A', 'A'],
    ['C', 'C', 'B', 'B'],
    ['D', 'A', 'A', 'D']
]
count_A = 0
count_B = 0
count_C = 0
count_D = 0

for row in matrix:
    for char in row:
        if char == 'A':
            count_A += 1
        elif char == 'B':
            count_B += 1
        elif char == 'C':
            count_C += 1
        elif char == 'D':
            count_D += 1

counts = [count_A, count_B, count_C, count_D]
chars = ['A', 'B', 'C', 'D']

for i in range(len(chars)):
    print(f"{chars[i]} 는 {counts[i]} 개 입니다.")
```

```python
#실습6
color_dict = {
        'red': '#FF0000',
        'blue': '#0000FF',
        'green': '#008000',
        'yellow': '#FFFF00',
        'orange': '#FFA500',
        'black': '#000000',
        'white': '#FFFFFF',
        'violet': '#EE82EE',
        'pink': '#FFC0CB',
        'lime': '#00FF00'
}
color_name = input("칼라명을 영문으로 입력하세요 : ")
if color_name in color_dict:
    print(f"{color_name} 칼라의 RGB 값은 {color_dict[color_name]} 입니다.")
else:
    print(f"{color_name} 칼라의 RGB 값을 찾을 수 없습니다")
```

```python
#실습7
weather_dict = {
        '목': (21, 27),
        '금': (18, 28),
        '토': (19, 30),
        '일': (20, 29)
    }

day = input("요일명을 한글로 입력하세요 : ")
if day in weather_dict:
    min_temp, max_temp = weather_dict[day]
    print(f"{day}요일의 최저온도는 {min_temp} 이고, 최고온도는 {max_temp}입니다.")
else:
    print("{day}요일의 정보를 찾을 수 없습니다.")
```

```python
#실습9
import random

set1 = set()
set2 = set()

while len(set1) < 10:
        set1.add(random.randint(1, 20))
while len(set2) < 10:
        set2.add(random.randint(1, 20))

print(f"집합 1 : {set1}")
print(f"집합 2 : {set2}")
print(f"두 집합에 모두 있는 데이터 : {set1 & set2}")
print(f"집합1 또는 집합2 에 있는 데이터 : {set1 | set2}")
print(f"집합1에는 있고 집합2에는 없는 데이터 : {set1 - set2}")
print(f"집합2에는 있고 집합1에는 없는 데이터 : {set2 - set1}")
print(f"집합1과 집합 2가 각자 가지고 있는 데이터 : {set1 ^ set2}")
```
