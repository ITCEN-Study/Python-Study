```python
# 실습4

for i in range(9,3,-1):
    if i % 2 == 0:
        print(f"{i} : 짝수")
    else:
        print(f"{i} : 홀수")
```

```python
# 실습5

import random

start = random.randint(1,10)
end = random.randint(30,40)

sum = 0
for i in range(start, end+1):
    if i % 2 == 0:
        sum += i
print(f"{start} 부터 {end}까지의 짝수의 합: {sum}")
```

```python
# 실습6

even_num = 0
odd_num = 0

for i in range(1,101):
    if i % 2 == 0:
        even_num += i
    else:
        odd_num += i

print("1 부터 100까지의 숫자들 중에서")
print(f"짝수의 합은 {even_num} 이고,")
print(f"홀수의 합은 {odd_num} 이다.")
```

```python
# 실습7

num = random.randint(5,10)

for i in range(1, num + 1):
    print(f"{i}->{i ** 2}")
```

```python
# 실습8

while True:
    pairNum1 = random.randint(1,6)
    pairNum2 = random.randint(1,6)

    if pairNum1 != pairNum2:
        if pairNum1 > pairNum2:
            print(f"{pairNum1} 이 {pairNum2} 보다 크다.")
        else:
            print(f"{pairNum1} 이 {pairNum2} 보다 작다.")
    else:
        print("게임 끝")
        break
```

```python
# 실습9

count = 0

while True:
    num = random.randint(0,30)
    count += 1

    if 1 <= num <= 26:
            char = chr(num + 64)
            print(f"{char}({num})")
    elif num == 0 or 27 <= num <= 30:
            break

print(f"수행횟수는 {count} 번입니다")
```

```python
# 실습10

while True:
        month_str = input("1~12 사이의 값을 입력하세요 : ")

        if month_str.isdigit():
            month = int(month_str)
            if 1 <= month <= 12:
                if month in [12, 1, 2]:
                    print(f"{month} 월은 겨울")
                elif month in [3, 4, 5]:
                    print(f"{month} 월은 봄")
                elif month in [6, 7, 8]:
                    print(f"{month} 월은 여름")
                elif month in [9, 10, 11]:
                    print(f"{month} 월은 가을")
            else:
                print("1~12 사이의 값을 입력하세요!")
                break
        else:
            print("1~12 사이의 값을 입력하세요!")
            break
```
