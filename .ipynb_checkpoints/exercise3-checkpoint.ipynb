```python
# 실습1

def print_my_nicname():
    print("포켓몬마스터 재혁")

def print_my_book():
    print("포켓몬스터 생태도감")
```

```python
# 실습1 호출 셀

for _ in range(3):
    print_my_nicname()

for _ in range(5):
    print_my_book()
```

```python
# 실습3

def expr(n1, n2, op):
    if op == "+":
        return n1 + n2
    elif op == "-":
        return n1 - n2
    elif op == "*":
        return n1 * n2
    elif op == "/":
        return n1 / n2
    else:
        return None
```

```python
# 실습3 호출 셀

res = expr(10, 5, "+")
if res is None:
    print("수행 불가")
else:
    print(f"연산결과 : {res}")
```

```python
# 실습4

def print_triangle1(num):
    if not (1 <= num <= 10):
        return
    for i in range(1, num + 1):
        print("*" * i)
```

```python
# 실습4 호출 셀

print_triangle1(2)
print_triangle1(5)

print_triangle1(12)
```

```python
# 실습 5

def print_gugudan(num):
    if type(num) != int:
        return
    if not (1 <= num <= 9):
        return
    for i in range(1,10):
        print(f"{num} x {i} = {num * i}")
```

```python
# 실습 5 호출 셀

print_gugudan(3)
print_gugudan(13)

print_gugudan("미출력")
```

```python
# 실습 6

def differ_two_value(val1, val2):
    if val1 > val2:
        return val1 - val2
    elif val1 < val2:
        return val2 - val1
    else:
        return 0
```

```python
# 실습 6 호출 셀

import random

for _ in range(5):
    x = random.randint(1, 30)
    y = random.randint(1, 30)
    w = differ_two_value(x, y)
    print(f"{x}와 {y}의 차는 {w} 입니다.")
```
