```python
# 실습1

listnum = [10, 5, 7, 21, 4, 8, 18]

# - 다음 형식으로 리스트의 값들을 출력한다. 10, 5, 7, 21, 4, 8, 18
for i in range(len(listnum)):
    if i == len(listnum) - 1:
        print(listnum[i])
    else:
        print(listnum[i], end=", ")
        
# - 리스트에 존재하는 짝수의 개수를 출력한다. 짝수의 갯수 : 4 개
even_count = 0
for num in listnum:
    if num % 2 == 0:
        even_count += 1
print(f"짝수의 갯수 : {even_count} 개")

# - listnum 에 저장된 값들 중에서 최솟값을 추출하여 다음과 같이 출력한다. 최솟값 : 4
min_val = listnum[0]
max_val = listnum[0]
for num in listnum:
    if num < min_val:
            min_val = num
    if num > max_val:
            max_val = num

print(f"최솟값 : {min_val}")
print(f"최댓값 : {max_val}")
```

```python
# 실습2
import random

# 비어있는 리스트 하나 생성
listnum = []

# 1~50 사이의 난수 10개 추출
for _ in range(10):
    listnum.append(random.randint(1, 50))

# 모든 값 출력
print("최초 리스트:", listnum)

# 첫 번째 데이터 출력
print("첫 번째 데이터:", listnum[0])

# 7. 마지막 데이터 출력
print("마지막 데이터:", listnum[-1])

# 8. 두 번째부터 여섯 번째 데이터 추출 출력 (인덱스 1부터 5까지)
print("2번째~6번째 데이터:", listnum[1:6])

# 9. 역순 출력
print("역순 리스트:", listnum[::-1])

# 10. 모두 출력
print("모든 데이터:", listnum[:])

# 11. 5번째 데이터 삭제 (인덱스 4)
del listnum[4]

# 12. 모두 출력
print("5번째 삭제 후:", listnum[:])

# 13. 2~3번째 데이터 삭제 (인덱스 1부터 2까지)
listnum[1:3] = []

# 14. 모두 출력
print("2~3번째 삭제 후:", listnum[:])
```

```python
# 실습3
def get_min_max(num_list):
    return min(num_list), max(num_list)
```

```python
# 실습 3 호출 셀

list_v = []
for _ in range(10):
    list_v.append(random.randint(1, 30))

min_v, max_v = get_min_max(list_v)

print(f"생성된 리스트 : {list_v}")
print(f"최솟값은 {min_v} 이고, 최댓값은 {max_v} 입니다.")
```

```python
# 실습4

def get_sum(tup):
    sum_val = 0
    count = 0
    for item in tup:
        if type(item) == int:
            sum_val += item
            count += 1
        avg_val = sum_val / count if count > 0 else 0.0
        return sum_val, avg_val, count
```

```python
t = (10, '가나', 30, True, 50, 'python')
result1, result2, result3 = get_sum(t)

print(f"합은 {result1} 이고 평균은 {result2} 그리고 데이터의 개수는 {result3} 입니다.")
```
