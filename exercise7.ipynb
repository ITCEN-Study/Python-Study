```python
#실습2
scores = {"Kim": 95, "Lee": 82, "Park": 67, "Choi": 58}

grades = {
    name: ('A' if score >= 90 else ('B' if score >= 80 else ('C' if score >= 70 else 'F')))
    for name, score in scores.items()
}
print(grades)
```

```python
#실습3
def createList(*args, type=1):
    data = args if len(args) > 0 else range(1, 31)

    if type == 2:
        return [x for x in data if x % 2 == 0]
    elif type == 3:
        return [x for x in data if x % 2 != 0]
    elif type == 4:
        return [x for x in data if x > 10]
    else:
        return [x for x in data]

print(createList(1, 2, 3, 11, 12, type=4))
print(createList(type=2))
```

```python
#실습4
def mycompre_dict(**kwargs):
    return {'my' + k: v for k, v in kwargs.items()}

print(mycompre_dict(a=1, b=2))
```

```python
#실습5
def make_comprelist(*nums, type=10):
    return [x * type for x in nums]

print(make_comprelist(1,2,3,4,5))                       # [10, 20, 30, 40, 50]
print(make_comprelist(1,2,3,4,5, type=3))               # [3, 6, 9, 12, 15]
print(make_comprelist(10, 20, 30, 40, 50, 60, type=2))  # [20, 40, 60, 80, 100, 120]
print(make_comprelist(1,2,3, type=100))                 # [100, 200, 300]
print(make_comprelist(1,1,1,1,1,1,1, type=0))           # [0, 0, 0, 0, 0, 0, 0]
```

```python
#실습8
import random

random_scores = [random.randint(0, 100) for _ in range(10)]
print("생성된 점수 리스트:", random_scores)

# (2) 딕셔너리 컴프리헨션을 활용한 pass/nopass 생성
# [방법 A] 점수 자체를 Key로 매핑할 때 (중복 점수가 있을 경우 덮어씌워짐)
result_dict = {score: ('pass' if score >= 60 else 'nopass') for score in random_scores}
print("결과 딕셔너리 (점수 키):", result_dict)
```
